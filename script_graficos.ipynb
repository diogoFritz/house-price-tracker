#### Visualização

```python
df = json_to_table("ficheiros\\propriedades_sapo_amadora_2025_09_21.json")

df = df[['preco', 'tamanho']]

print(df.count())
```

```python
def plot_scatter(df):
    plt.figure(figsize=(8, 6))

    # Adiciona o gráfico de dispersão
    sns.scatterplot(data=df, x='metros', y='preco', marker="X", s=200)

    # Altere para o estilo desejado
    sns.set_style("darkgrid")  

    # Configurar Escala
    plt.xlim(25, 130)
    plt.xticks(range(25, 121, 10))
    plt.ylim(120_000, 400_000)
    plt.yticks(range(120_000, 401_000, 20_000))

    # Definir Eixos e Título
    plt.xlabel('Metros')
    plt.ylabel('Preço')
    plt.title('Relação entre Metros e Preço')
    plt.grid(True)
    plt.show()

plot_scatter(df)
```

Visualização - Heatmap

```python
# Adiciona o heatmap de densidade
sns.kdeplot(
    data=df,
    x="metros",
    y="preco",
    fill=True,
    cmap="Reds",
    alpha=0.4,
    thresh=0.05
)
```
